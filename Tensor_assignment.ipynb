```python
import torch
import random

torch.manual_seed(1234)
tensor_C = torch.rand(size=(2,3))

torch.manual_seed(1234)
tensor_D = torch.rand(size=(2,3))

tensor_C_GPU = tensor_C.to(device = "cuda")
tensor_D_GPU = tensor_D.to(device = "cuda")

tensor_C_GPU, tensor_D_GPU
```

```python
tensor_CD_GPU = torch.matmul(tensor_C_GPU, tensor_D_GPU.T)
tensor_CD_GPU
```

```python
tensor_CD_GPU.max(), tensor_CD_GPU.min()
```

```python
tensor_CD_GPU_argmax = tensor_CD_GPU.argmax()
tensor_CD_GPU_argmin = tensor_CD_GPU.argmin()
tensor_CD_GPU_argmax, tensor_CD_GPU_argmin
```
